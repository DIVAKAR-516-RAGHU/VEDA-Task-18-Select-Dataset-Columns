## Task 18: Select Dataset Columns

### Objective
Learn how to access and select specific rows and columns from a Pandas DataFrame using `loc` and `iloc`.

### Tools Used
- Python
- Pandas
- Jupyter Notebook

### Dataset
Iris Dataset

## 1. Import Pandas

Pandas is a Python library used for data manipulation and analysis.

```python
import pandas as pd
```

## 2. Load the Iris Dataset

The Iris dataset contains measurements of iris flowers.

It has the following features:
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width
- Species

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

df = pd.read_csv(url, names=columns)

df.head()
```

## 3. Explore the Dataset

First, we check the shape and column names of the DataFrame.

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

## 4. Select a Single Column Using `loc`

The `loc` accessor is label-based.

The following code selects the `sepal_length` column from all rows.

```python
df.loc[:, "sepal_length"]
```

## 5. Select Multiple Columns Using `loc`

Multiple columns can be selected by providing their column names inside a list.

```python
df.loc[:, ["sepal_length", "petal_length", "species"]]
```

## 6. Select Specific Rows and Columns Using `loc`

`loc` can be used to select both specific rows and specific columns.

Here, we select rows 0 to 4 and three columns:
- Sepal Length
- Petal Length
- Species

```python
df.loc[0:4, ["sepal_length", "petal_length", "species"]]
```

## 7. Select a Single Column Using `iloc`

The `iloc` accessor is position-based.

The following code selects the first column using its integer position `0`.

```python
df.iloc[:, 0]
```

## 8. Select Multiple Columns Using `iloc`

We can select multiple columns by specifying their integer positions.

Column positions:
- 0 → sepal_length
- 1 → sepal_width
- 2 → petal_length
- 3 → petal_width
- 4 → species

```python
df.iloc[:, [0, 2, 4]]
```

## 9. Select Specific Rows and Columns Using `iloc`

The following code selects the first five rows and columns at positions 0, 2, and 4.

```python
df.iloc[0:5, [0, 2, 4]]
```

## 10. Difference Between `loc` and `iloc`

| Method | Selection Type | Example |
|---|---|---|
| `loc` | Label-based | `df.loc[:, "sepal_length"]` |
| `iloc` | Integer position-based | `df.iloc[:, 0]` |

### Key Difference

- `loc` uses row and column labels.
- `iloc` uses integer positions.
- `loc` is useful when working with meaningful labels.
- `iloc` is useful when selecting data based on its position.

## 11. Select Final Records

Now we select the first 10 records and three specific columns using `loc`.

```python
selected_data = df.loc[
    0:9,
    ["sepal_length", "petal_width", "species"]
]

selected_data
```
